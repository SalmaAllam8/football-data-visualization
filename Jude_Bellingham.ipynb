```python
club_standard_stats = pd.read_csv(r"C:\Users\salma\Downloads\Club_Stats.csv")
domestic_league= pd.read_csv(r"C:\Users\salma\Downloads\Domestic_leagues.csv")
domestic_cup= pd.read_csv(r"C:\Users\salma\Downloads\Domestic_cups.csv")
international_cups = pd.read_csv(r"C:\Users\salma\Downloads\International_cups.csv")
national_team = pd.read_csv(r"C:\Users\salma\Downloads\National_Team.csv")

#shooting

shooting_club_standard_stats = pd.read_csv(r"C:\Users\salma\Downloads\Shooting_Standard_Stats.csv")
shooting_domestic_league = pd.read_csv(r"C:\Users\salma\Downloads\League_Shooting_Standard_Stats_and_Summary.csv")
shooting_domestic_cup = pd.read_csv(r"C:\Users\salma\Downloads\Domestic_Cups_Shooting_Standard_Stats_and_Summary.csv")
shooting_international_cups = pd.read_csv(r"C:\Users\salma\Downloads\European_International_Cups_Shooting_Standard_Stats.csv")
shooting_national_team = pd.read_csv(r"C:\Users\salma\Downloads\International_England_Shooting_Standard_Stats.csv")
```

```python
club_standard_stats = club_standard_stats.copy()
domestic_league = domestic_league.copy()
domestic_cup = domestic_cup.copy()
international_cups = international_cups.copy()
national_team = national_team.copy()

club_standard_stats["Competition_Type"] = "Club"
domestic_league["Competition_Type"] = "Domestic League"
domestic_cup["Competition_Type"] = "Domestic Cup"
international_cups["Competition_Type"] = "International Cup"
national_team["Competition_Type"] = "National Team"
```

```python
shooting_club_standard_stats["Competition_Type"] = "Club"
shooting_domestic_league["Competition_Type"] = "Domestic League"
shooting_domestic_cup["Competition_Type"] = "Domestic Cup"
shooting_international_cups["Competition_Type"] = "International Cup"
shooting_national_team["Competition_Type"] = "National Team"
```

```python
standard_stats = pd.concat(
    [
        club_standard_stats,
        domestic_league,
        domestic_cup,
        international_cups,
        national_team
    ],
    ignore_index=True
)
```

```python
numeric_cols = [
    "Age", "MP", "Starts", "Min", "90s",
    "Gls", "Ast", "G+A", "G-PK", "PK", "PKatt",
    "CrdY", "CrdR",
    "Gls_90", "Ast_90", "G+A_90",
    "G-PK_90", "G+A-PK_90"
]

standard_stats[numeric_cols] = standard_stats[numeric_cols].apply(
    pd.to_numeric,
    errors="coerce"
)
```

```python
shooting_numeric_cols = [
    "Age", "90s", "Gls", "Sh", "SoT",
    "SoT%", "Sh_90", "SoT_90",
    "G_Sh", "G_SoT", "PK", "PKatt"
]

shooting[shooting_numeric_cols] = shooting[shooting_numeric_cols].apply(
    pd.to_numeric,
    errors="coerce"
)
```

```python
player_stats = standard_stats.merge(
    shooting,
    on=["Season", "Squad", "Comp", "Competition_Type"],
    how="left",
    suffixes=("", "_shooting")
)
```

### Start percentage

```python
standard_stats["Start_Rate"] = (
    standard_stats["Starts"] / standard_stats["MP"]
)
```

### Minutes per match

```python
standard_stats["Minutes_per_Match"] = (
    standard_stats["Min"] / standard_stats["MP"]
)
```

### Penalty conversion

```python
standard_stats["Penalty_Conversion"] = (
    standard_stats["PK"] /
    standard_stats["PKatt"].replace(0, pd.NA)
)
```
