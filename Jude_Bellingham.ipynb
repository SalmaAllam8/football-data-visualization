```python


fig, ax = plt.subplots(figsize=(12, 6), facecolor="#0e1117")
ax.set_facecolor("#0e1117")

bars = ax.bar(
    career_overview["year"].astype(str),
    career_overview["Minutes"],
    color="#2ee6a6",
    width=0.6
)

# Value labels on top of each bar
for bar, val in zip(bars, career_overview["Minutes"]):
    ax.annotate(
        f"{val:,.0f}",
        xy=(bar.get_x() + bar.get_width() / 2, val),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center",
        color="white",
        fontsize=9,
        fontweight="bold"
    )

# Reference line: career average minutes per season
avg_minutes = career_overview["Minutes"].mean()
ax.axhline(avg_minutes, color="#4ea8ff", linestyle="--", linewidth=1.2, alpha=0.7)
ax.text(
    len(career_overview) - 0.5, avg_minutes,
    f"  Avg: {avg_minutes:,.0f}",
    color="#4ea8ff", fontsize=10, va="bottom", ha="right", fontweight="bold"
)

# Styling
ax.set_title("Minutes Played Across the Player's Career", color="white",
             fontsize=16, fontweight="bold", loc="left", pad=15)
ax.set_xlabel("Year", color="#aaaaaa", fontsize=11)
ax.set_ylabel("Minutes", color="#aaaaaa", fontsize=11)
ax.tick_params(colors="#aaaaaa", labelsize=10)
plt.xticks(rotation=45, ha="right")
for spine in ax.spines.values():
    spine.set_visible(False)
ax.grid(axis="y", color="white", alpha=0.1)

plt.tight_layout()
plt.show()
```
