```python
standard_stats['year'] = standard_stats['Season'].str.split('-').str[-1]

# Optional: convert to integer format
standard_stats['year'] = standard_stats['year'].astype(int)
```

```python
career_overview = (
    standard_stats
    .groupby(["year", "Age"], as_index=False)
    .agg(
        Clubs=("Squad", "nunique"),
        Competitions=("Comp", "nunique"),
        Minutes=("Min", "sum")
    )
)

career_overview
```

```python


fig, ax = plt.subplots(figsize=(12, 6), facecolor="#0e1117")
ax.set_facecolor("#0e1117")

bars = ax.bar(
    career_overview["year"].astype(str),
    career_overview["Minutes"],
    color="#2ee6a6",
    width=0.6
)

# Value labels on top of each bar
for bar, val in zip(bars, career_overview["Minutes"]):
    ax.annotate(
        f"{val:,.0f}",
        xy=(bar.get_x() + bar.get_width() / 2, val),
        xytext=(0, 5),
        textcoords="offset points",
        ha="center",
        color="white",
        fontsize=9,
        fontweight="bold"
    )

# Reference line: career average minutes per season
avg_minutes = career_overview["Minutes"].mean()
ax.axhline(avg_minutes, color="#4ea8ff", linestyle="--", linewidth=1.2, alpha=0.7)
ax.text(
    len(career_overview) - 0.5, avg_minutes,
    f"  Avg: {avg_minutes:,.0f}",
    color="#4ea8ff", fontsize=10, va="bottom", ha="right", fontweight="bold"
)

# Styling
ax.set_title("Minutes Played Across the Player's Career", color="white",
             fontsize=16, fontweight="bold", loc="left", pad=15)
ax.set_xlabel("Year", color="#aaaaaa", fontsize=11)
ax.set_ylabel("Minutes", color="#aaaaaa", fontsize=11)
ax.tick_params(colors="#aaaaaa", labelsize=10)
plt.xticks(rotation=45, ha="right")
for spine in ax.spines.values():
    spine.set_visible(False)
ax.grid(axis="y", color="white", alpha=0.1)

plt.tight_layout()
plt.show()
```

# Player performance over time

## Goals per 90

```python
goals_over_time = (
    club_standard_stats[["Season", "Age", "Gls_90"]]
    .drop_duplicates()
    .sort_values("Season")
)

goals_over_time
```

```python
club_standard_stats.groupby("Season").size()
```

```python
fig, ax = plt.subplots(figsize=(11, 6), facecolor="#0e1117")
ax.set_facecolor("#0e1117")

x = goals_over_time["Season"].astype(str)
y = goals_over_time["Gls_90"]

# Line + markers
ax.plot(x, y, marker="o", markersize=8, color="#2ee6a6", linewidth=2.2,
        markerfacecolor="#2ee6a6", markeredgecolor="white", markeredgewidth=1.2, zorder=3)

# Fill under the line for visual weight
ax.fill_between(x, y, color="#2ee6a6", alpha=0.08, zorder=1)

# Value labels above each point
for xi, yi in zip(x, y):
    ax.annotate(
        f"{yi:.2f}",
        xy=(xi, yi),
        xytext=(0, 10),
        textcoords="offset points",
        ha="center",
        color="white",
        fontsize=9,
        fontweight="bold"
    )

# Career average reference line
avg_val = y.mean()
ax.axhline(avg_val, color="#4ea8ff", linestyle="--", linewidth=1.2, alpha=0.6, zorder=2)
ax.text(
    1, avg_val+0.01,
    f"  Career avg: {avg_val:.2f}",
    color="#4ea8ff", fontsize=10, va="bottom", ha="right"
)

# Styling
ax.set_title("Goals per 90 Across Jude Bellingham's Career", color="white",
             fontsize=16, fontweight="bold", loc="left", pad=15)
ax.set_xlabel("Season", color="#aaaaaa", fontsize=11)
ax.set_ylabel("Goals per 90", color="#aaaaaa", fontsize=11)
ax.tick_params(colors="#aaaaaa", labelsize=10)
plt.xticks(rotation=45, ha="right")
for spine in ax.spines.values():
    spine.set_visible(False)
ax.grid(axis="y", color="white", alpha=0.1)

plt.tight_layout()
plt.show()
```

### highest and lowest Season

```python
best_season = goals_over_time.loc[
    goals_over_time["Gls_90"].idxmax()
]

lowest_season = goals_over_time.loc[
    goals_over_time["Gls_90"].idxmin()
]

print("Highest:", best_season)
print("Lowest:", lowest_season)
```

###  the year-to-year change

```python
goals_over_time["Gls_90_change"] = (
    goals_over_time["Gls_90"].diff()
)

```
